```python
import numpy as np
```

```python
data = np.array([
    [2, 3],
    [4, 5],
    [6, 7],
    [8, 9],
    [1, 2]
])

print(data)
```

```python
new_entry = np.array([5, 6])

print("New Entry:", new_entry)
```

```python
distances = []

for point in data:
    distance = np.sqrt(np.sum((new_entry - point) ** 2))
    distances.append(distance)

print("Distances:", distances)
```

```python
for i in range(len(data)):
    print(
        "Existing Point:", data[i],
        "Distance:", distances[i]
    )
```

```python
nearest_index = np.argmin(distances)

print("New Entry:", new_entry)
print("Nearest Point:", data[nearest_index])
print("Minimum Distance:", distances[nearest_index])
```
